# QQQ Equity Curves + Performance (Python)

This notebook recreates the QQQ Rmd workflow in Python with updates:
- includes `Equal-Weight Monthly`
- includes `Markowitz Min Variance`
- computes `IR3 = ARC^3 / (ASD * MD * MLD)`
- exports 3 panel plots to `output/QQQ`
- prints a LaTeX-ready panel table in cell output

```python
from pathlib import Path
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# High-quality vector PDF output settings
plt.rcParams["pdf.fonttype"] = 42
plt.rcParams["ps.fonttype"] = 42
plt.rcParams["savefig.dpi"] = 600


# -------------------------
# Config
# -------------------------
MARKET = "QQQ"
ANNUALIZATION = 252

PANELS = {
    "Panel A: RL with LSTM and Cash Allowed": {
        "rl_runs": {
            "LSTM_1": "QQQ_LSTM_DAILY_1",
            "LSTM_2": "QQQ_LSTM_DAILY_2",
        },
        "plot_file": "fig_1-1.pdf",
    },
    "Panel B: RL with LSTM and Cash Not-Allowed": {
        "rl_runs": {
            "LSTM_NC_1": "QQQ_LSTM_NC_DAILY",
            "LSTM_NC_2": "QQQ_LSTM_NC_DAILY_2",
        },
        "plot_file": "fig_2-1.pdf",
    },
    "Panel C: RL with Transformers": {
        "rl_runs": {
            "TRANSFORMERS": "QQQ_TRANSFORMERS_DAILY_1",
        },
        "plot_file": "fig_3-1.pdf",
    },
}

BASELINE_RUN = "QQQ_LSTM_DAILY_1"
BASELINE_STRATEGIES = {
    "Buy & Hold": "QQQ Buy-and-Hold",
    "Momentum Top-20": "Momentum Top-20%",
    "Equal-Weight Monthly": "Equal-Weight Monthly",
}
MARKOWITZ_LABEL = "Markowitz Min Variance"
MARKOWITZ_COL = "QQQ_Min Variance"

PLOT_COLORS = {
    "Buy & Hold": "black",
    "Momentum Top-20": "darkgreen",
    "Equal-Weight Monthly": "dodgerblue",
    MARKOWITZ_LABEL: "orange",
    "LSTM_1": "darkred",
    "LSTM_2": "purple",
    "LSTM_NC_1": "darkred",
    "LSTM_NC_2": "purple",
    "TRANSFORMERS": "darkred",
}


# -------------------------
# Paths and loaders
# -------------------------
def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / ".git").exists() and (candidate / "Results_Daily").exists():
            return candidate
    raise FileNotFoundError("Could not locate repository root.")


REPO_ROOT = find_repo_root(Path.cwd().resolve())
RESULTS_ROOT = REPO_ROOT / "Results_Daily" / MARKET
MARKOWITZ_FILE = REPO_ROOT / "v1_Changes_Calculations" / "Benchmark_Markowitz" / "markowitz_daily_returns_QQQ.csv"
OUT_DIR = REPO_ROOT / "v1_Changes_Calculations" / "New_Equity_Curves_Performance" / "output" / "QQQ"
OUT_DIR.mkdir(parents=True, exist_ok=True)


def load_returns(run_folder: str, file_name: str) -> pd.DataFrame:
    p = RESULTS_ROOT / run_folder / file_name
    df = pd.read_csv(p)
    if "date" not in df.columns:
        first = df.columns[0]
        df = df.rename(columns={first: "date"})
    df["date"] = pd.to_datetime(df["date"])
    return df.sort_values("date")


def load_rl_series(run_folder: str) -> pd.Series:
    df = load_returns(run_folder, "rl_daily_returns_oos.csv")
    return df.set_index("date")["RL Agent"]


def load_baseline_series() -> dict:
    df = load_returns(BASELINE_RUN, "daily_returns_oos.csv").set_index("date")
    out = {}
    for nice, raw in BASELINE_STRATEGIES.items():
        out[nice] = df[raw]
    return out


def load_markowitz_series() -> pd.Series:
    df = pd.read_csv(MARKOWITZ_FILE)
    df["date"] = pd.to_datetime(df["date"])
    return df.set_index("date")[MARKOWITZ_COL]


def estimate_markowitz_adt_pct_from_tc_diff(
    tc_bps: float = 2.0,
    variant: str = "min_variance",
) -> float:
    """
    Estimate Markowitz ADT (%) by rerunning WFO twice (tc=0 and tc=tc_bps)
    and backing out implied turnover from daily net-return differences.
    """
    import sys

    code_dir = REPO_ROOT / "Code"
    mkz_dir = REPO_ROOT / "v1_Changes_Calculations" / "Benchmark_Markowitz"

    if str(code_dir) not in sys.path:
        sys.path.insert(0, str(code_dir))
    if str(mkz_dir) not in sys.path:
        sys.path.insert(0, str(mkz_dir))

    from functions.data_pipeline_daily import build_dataset
    from markowitz import run_markowitz_wfo

    dataset = build_dataset(str(REPO_ROOT / "Data" / MARKET / "Outputs" / "Data"))

    common_kwargs = dict(
        variant=variant,
        train_months=60,
        val_months=12,
        test_months=12,
        step_months=12,
        rebalance_freq_days=21,
        max_weight=0.10,
        top_k=20,
        annualization=ANNUALIZATION,
        verbose=False,
    )

    r0 = run_markowitz_wfo(dataset, transaction_cost_bps=0.0, **common_kwargs)["daily_returns"]
    rt = run_markowitz_wfo(dataset, transaction_cost_bps=tc_bps, **common_kwargs)["daily_returns"]

    idx = r0.index.intersection(rt.index)
    if len(idx) == 0:
        return np.nan

    tc_rate = tc_bps / 10_000.0
    implied_turnover = (r0.loc[idx] - rt.loc[idx]) / tc_rate

    # numerical stability: tiny negatives can appear due to floating point
    implied_turnover = implied_turnover.clip(lower=0)

    adt_pct = float(implied_turnover.mean() * 100.0)
    return adt_pct


def load_baseline_adt() -> dict:
    p = RESULTS_ROOT / BASELINE_RUN / "performance_metrics_oos.csv"
    df = pd.read_csv(p)
    m = dict(zip(df["Strategy"], df["Avg Daily Turnover (%)"]))
    return {nice: float(m.get(raw, np.nan)) for nice, raw in BASELINE_STRATEGIES.items()}


def load_rl_adt(run_folder: str) -> float:
    p = RESULTS_ROOT / run_folder / "rl_performance_metrics.csv"
    df = pd.read_csv(p)
    if "Unnamed: 0" in df.columns:
        df = df.rename(columns={"Unnamed: 0": "Strategy"})
    rl_row = df.loc[df["Strategy"] == "RL Agent"]
    return float(rl_row["Avg Daily Turnover (%)"].iloc[0]) if not rl_row.empty else np.nan


# -------------------------
# Metrics
# -------------------------
def equity_to_returns(tab: np.ndarray) -> np.ndarray:
    tab = np.asarray(tab, dtype=float)
    return (tab[1:] / tab[:-1]) - 1.0


def absolute_return(tab: np.ndarray) -> float:
    ret = equity_to_returns(tab)
    return float((np.prod(1.0 + ret) - 1.0) * 100.0) if len(ret) else 0.0


def arc(tab: np.ndarray, annualization: int = 252) -> float:
    tab = np.asarray(tab, dtype=float)
    ret = equity_to_returns(tab)
    if len(tab) <= 1:
        return 0.0
    a_rtn = np.prod(1.0 + ret[:-1]) if len(ret) > 1 else 1.0
    if a_rtn <= 0:
        return 0.0
    return float(100.0 * (math.pow(a_rtn, annualization / len(tab)) - 1.0))


def asd(tab: np.ndarray, annualization: int = 252) -> float:
    ret = equity_to_returns(tab)
    return float(np.sqrt(annualization) * np.std(ret) * 100.0) if len(ret) else 0.0


def mdd(tab: np.ndarray) -> float:
    ret = equity_to_returns(tab)
    if len(ret) == 0:
        return 0.0
    cum = np.cumprod(1.0 + ret)
    cmax = np.maximum.accumulate(cum)
    dd = (cmax - cum) / cmax
    return float(np.max(dd) * 100.0)


def mld(tab: np.ndarray, annualization: int = 252) -> float:
    temp = np.asarray(tab, dtype=float)
    if len(temp) == 0:
        return 1.0
    i = int(np.argmax(np.maximum.accumulate(temp) - temp))
    if i == 0:
        return len(temp) / annualization
    j = int(np.argmax(temp[:i]))
    mld_end = -1
    for k in range(i, len(temp)):
        if (temp[k - 1] < temp[j]) and (temp[j] < temp[k]):
            mld_end = k
            break
    if mld_end == -1:
        mld_end = len(temp)
    return float(abs(mld_end - j) / annualization)


def ir1(arc_v: float, asd_v: float) -> float:
    return float(max(arc_v / asd_v, 0.0)) if asd_v != 0 else 0.0


def ir2(arc_v: float, asd_v: float, mdd_v: float) -> float:
    denom = asd_v * mdd_v
    if denom == 0:
        return 0.0
    numer = (arc_v ** 2) * (0 if arc_v == 0 else int(abs(arc_v) / arc_v))
    return float(max(numer / denom, 0.0))


def ir3(arc_v: float, asd_v: float, mdd_v: float, mld_v: float) -> float:
    denom = asd_v * mdd_v * mld_v
    if denom == 0:
        return 0.0
    return float((arc_v ** 3) / denom)


def sharpe_from_returns(r: np.ndarray, annualization: int = 252) -> float:
    if len(r) <= 1 or np.std(r) <= 1e-12:
        return 0.0
    return float(np.mean(r) / np.std(r) * np.sqrt(annualization))


def compute_metrics_from_returns(returns: pd.Series, adt: float = np.nan) -> dict:
    r = returns.dropna().to_numpy(dtype=float)
    eq = np.concatenate(([1.0], np.cumprod(1.0 + r)))

    ar_v = absolute_return(eq)
    arc_v = arc(eq, ANNUALIZATION)
    asd_v = asd(eq, ANNUALIZATION)
    md_v = mdd(eq)
    mld_v = mld(eq, ANNUALIZATION)

    return {
        "AR (%)": ar_v,
        "ARC (%)": arc_v,
        "ASD (%)": asd_v,
        "MD (%)": md_v,
        "MLD": mld_v,
        "IR1": ir1(arc_v, asd_v) * 100.0,
        "IR2": ir2(arc_v, asd_v, md_v) * 100.0,
        "IR3": ir3(arc_v, asd_v, md_v, mld_v),
        "Sharpe": sharpe_from_returns(r, ANNUALIZATION),
        "ADT": float(adt) if pd.notna(adt) else np.nan,
    }


def make_equity(returns: pd.Series) -> pd.Series:
    s = returns.dropna().sort_index()
    return (1.0 + s).cumprod()


# -------------------------
# Build strategy series
# -------------------------
baseline_series = load_baseline_series()
baseline_adt = load_baseline_adt()
markowitz_series = load_markowitz_series()
markowitz_adt_pct = estimate_markowitz_adt_pct_from_tc_diff(tc_bps=2.0, variant="min_variance")
print(f"Estimated {MARKOWITZ_LABEL} ADT (%): {markowitz_adt_pct:.3f}")


def collect_panel_series(panel_cfg: dict) -> dict:
    sdict = {
        "Buy & Hold": baseline_series["Buy & Hold"],
        "Momentum Top-20": baseline_series["Momentum Top-20"],
        "Equal-Weight Monthly": baseline_series["Equal-Weight Monthly"],
        MARKOWITZ_LABEL: markowitz_series,
    }
    for rl_label, run_folder in panel_cfg["rl_runs"].items():
        sdict[rl_label] = load_rl_series(run_folder)
    return sdict


# -------------------------
# Plot panels (unified y-axis max)
# -------------------------
panel_aligned = {}
global_ymax = -np.inf
global_ymin = np.inf

for panel_name, panel_cfg in PANELS.items():
    sdict = collect_panel_series(panel_cfg)
    aligned = pd.concat(sdict, axis=1, join="inner").dropna()
    panel_aligned[panel_name] = aligned

    for col in aligned.columns:
        eq = make_equity(aligned[col])
        if len(eq):
            global_ymax = max(global_ymax, float(eq.max()))
            global_ymin = min(global_ymin, float(eq.min()))

if not np.isfinite(global_ymax):
    global_ymax = 1.0
if not np.isfinite(global_ymin):
    global_ymin = 0.0

for panel_name, panel_cfg in PANELS.items():
    aligned = panel_aligned[panel_name]

    fig, ax = plt.subplots(figsize=(12, 6))
    for col in aligned.columns:
        eq = make_equity(aligned[col])
        ax.plot(eq.index, eq.values, label=col, color=PLOT_COLORS.get(col, None), linewidth=1.6)

    ax.set_title(panel_name, fontsize=13)
    ax.set_xlabel("Date")
    ax.set_ylabel("Equity")
    ax.set_ylim(bottom=global_ymin, top=global_ymax)
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper left", frameon=True, fontsize=9)
    fig.tight_layout()

    out_path = OUT_DIR / panel_cfg["plot_file"]
    fig.savefig(out_path, dpi=600, bbox_inches="tight", facecolor="white", metadata={"Creator": "Matplotlib"})
    plt.show()
    print(f"Saved plot: {out_path}")


# -------------------------
# Performance table (panel style)
# -------------------------
metric_dirs = {
    "AR (%)": "max",
    "ARC (%)": "max",
    "ASD (%)": "min",
    "MD (%)": "min",
    "MLD": "min",
    "IR1": "max",
    "IR2": "max",
    "IR3": "max",
    "Sharpe": "max",
    "ADT": "min",  # ignore zeros when selecting best
}


def pick_best_mask(values: pd.Series, direction: str, ignore_zeros: bool = False) -> pd.Series:
    x = pd.to_numeric(values, errors="coerce")
    work = x.copy()
    if ignore_zeros:
        work = work[(work != 0) & work.notna()]
    if work.empty:
        return pd.Series(False, index=values.index)
    best = work.max() if direction == "max" else work.min()
    return x.eq(best)


def fmt_value(col: str, v: float) -> str:
    if pd.isna(v):
        return "-"
    if col == "AR (%)":
        return f"{v:.1f}"
    if col in {"ARC (%)", "ASD (%)", "MD (%)"}:
        return f"{v:.2f}"
    if col == "MLD":
        return f"{v:.3f}"
    if col in {"IR1", "IR2", "IR3", "Sharpe"}:
        return f"{v:.4f}"
    if col == "ADT":
        return f"{v:.3f}"
    return f"{v:.4f}"


def build_panel_perf(panel_name: str, panel_cfg: dict) -> pd.DataFrame:
    rows = []

    # Benchmarks
    for strat in ["Buy & Hold", "Momentum Top-20", "Equal-Weight Monthly"]:
        m = compute_metrics_from_returns(baseline_series[strat], adt=baseline_adt.get(strat, np.nan))
        rows.append({"Panel": panel_name, "Strategy": strat, **m})

    # Markowitz Min Variance
    m_mkz = compute_metrics_from_returns(markowitz_series, adt=markowitz_adt_pct)
    rows.append({"Panel": panel_name, "Strategy": MARKOWITZ_LABEL, **m_mkz})

    # RL strategies
    for rl_label, run_folder in panel_cfg["rl_runs"].items():
        m_rl = compute_metrics_from_returns(load_rl_series(run_folder), adt=load_rl_adt(run_folder))
        rows.append({"Panel": panel_name, "Strategy": rl_label, **m_rl})

    return pd.DataFrame(rows)


panel_tables = [build_panel_perf(name, cfg) for name, cfg in PANELS.items()]
perf = pd.concat(panel_tables, ignore_index=True)

# Display numeric table
print("\nNumeric performance table with IR3")
display(perf)


# -------------------------
# Build LaTeX-style table string
# -------------------------
panel_order = list(PANELS.keys())
metric_cols = ["AR (%)", "ARC (%)", "ASD (%)", "MD (%)", "MLD", "IR1", "IR2", "IR3", "Sharpe", "ADT"]

lines = []
lines.append(r"\begin{table}[htbp]")
lines.append(r"\centering")
lines.append(r"\scriptsize")
lines.append(r"\setlength{\tabcolsep}{3pt}")
lines.append(r"\renewcommand{\arraystretch}{0.95}")
lines.append(r"\resizebox{\textwidth}{!}{%")
lines.append(r"\begin{tabular}{lcccccccccc}")
lines.append("")
lines.append(r"\toprule")
lines.append(r"\textbf{Strategy} & \textbf{AR (\%)} & \textbf{ARC (\%)} & \textbf{ASD (\%)} & \textbf{MD (\%)} & \textbf{MLD} & \textbf{IR1} & \textbf{IR2} & \textbf{IR3} & \textbf{Sharpe} & \textbf{ADT} \\")
lines.append(r"\midrule")

for p_name in panel_order:
    p_df = perf.loc[perf["Panel"] == p_name].copy().reset_index(drop=True)
    lines.append("")
    lines.append(rf"\multicolumn{{11}}{{l}}{{\textbf{{{p_name}}}}} \\")
    lines.append(r"\midrule")

    best_masks = {}
    for mc in metric_cols:
        best_masks[mc] = pick_best_mask(
            p_df[mc],
            metric_dirs[mc],
            ignore_zeros=(mc == "ADT"),
        )

    for i, row in p_df.iterrows():
        strategy = str(row["Strategy"]).replace("&", r"\&").replace("_", r"\_")
        vals = []
        for mc in metric_cols:
            txt = fmt_value(mc, row[mc])
            if best_masks[mc].iloc[i] and txt != "-":
                txt = rf"\textbf{{{txt}}}"
            vals.append(txt)

        lines.append(strategy + " & " + " & ".join(vals) + r" \\")

    lines.append(r"\addlinespace")

lines.append(r"\bottomrule")
lines.append(r"\end{tabular}%")
lines.append(r"}")
lines.append(r"\end{table}")

latex_table = "\n".join(lines)

print("\nLaTeX table (copy/paste):\n")
print(latex_table)

print(f"\nAll plots saved under: {OUT_DIR}")
```
